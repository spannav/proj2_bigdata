## 1. ปัญหาและคำถามที่สนใจ
**ปัญหา/คำถามที่สนใจคืออะไร**

ในมหานครนิวยอร์ก (New York City) ความต้องการใช้บริการรถแท็กซี่มีความผันผวนอย่างรุนแรงตามช่วงเวลาของวัน วันในสัปดาห์ โซนพื้นที่ และโดยเฉพาะอย่างยิ่ง **การเปลี่ยนแปลงตามฤดูกาลทั้ง 4 ฤดูกาล (ฤดูหนาว ฤดูใบไม้ผลิ ฤดูร้อน และฤดูใบไม้ร่วง)** เช่น ในฤดูหนาวมักมีพายุหิมะและอุณหภูมิติดลบส่งผลให้ความต้องการรถเพิ่มขึ้นฉับพลันในระยะทางสั้น ขณะที่ฤดูร้อนมีนักท่องเที่ยวหนาแน่นในโซนสวนสาธารณะและสถานที่สำคัญ ส่วนฤดูใบไม้ร่วงมีกิจกรรมทางธุรกิจ การเปิดภาคเรียน และการประชุมสมัชชาใหญ่สหประชาชาติ (UN General Assembly) ที่ทำให้การจราจรหนาแน่นเป็นพิเศษ

**คำถามหลักของงานวิจัยนี้คือ:**
> *"เราสามารถพัฒนาระบบการเรียนรู้ของเครื่อง (Machine Learning Pipeline) เพื่อพยากรณ์จำนวนความต้องการเรียกรถแท็กซี่ (`demand_count`) ล่วงหน้า 1 ชั่วโมงในแต่ละโซนพื้นที่รับผู้โดยสาร (`PULocationID`) ได้แม่นยำเพียงใดตลอดทั้ง 4 ฤดูกาลของปี และปัจจัยด้านเวลา วัฏจักรย้อนหลัง ฤดูกาล หรือพื้นที่ มีอิทธิพลต่อความแม่นยำในการพยากรณ์มากที่สุด?"*

**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้**

1. **ความสำคัญเชิงเศรษฐกิจและสิ่งแวดล้อม:**
   * ปัญหาความไม่สมดุลระหว่างอุปสงค์และอุปทาน (Demand-Supply Mismatch) นำไปสู่ "การวิ่งรถเปล่า" (Deadheading หรือ Cruising without passengers) ซึ่งก่อให้เกิดการปล่อยก๊าซคาร์บอนไดออกไซด์ สิ้นเปลืองน้ำมันเชื้อเพลิง และเพิ่มความหนาแน่นของการจราจรในเขตแมนฮัตตัน
   * ผู้โดยสารต้องเผชิญกับเวลารอรถที่ยาวนาน (Excessive Wait Times) หรือไม่สามารถหารถได้ในชั่วโมงเร่งด่วนและช่วงสภาพอากาศเลวร้าย
2. **กลุ่มผู้ได้รับประโยชน์โดยตรง (Target Stakeholders):**
   * **ศูนย์บริหารจัดการกลุ่มรถ (Fleet Operators & Dispatchers):** ใช้โมเดลพยากรณ์ล่วงหน้า 1 ชั่วโมงเพื่อจัดสรรกองยานพาหนะ (Proactive Fleet Rebalancing) เคลื่อนย้ายรถไปยังโซนที่มีแนวโน้มความต้องการสูงล่วงหน้า
   * **ผู้ขับขี่รถแท็กซี่ (Taxi Drivers):** เพิ่มอัตราการมีผู้โดยสารบนรถ (Utilization Rate) ลดระยะเวลาวิ่งหาผู้โดยสารอย่างไร้จุดหมาย และเพิ่มรายได้ต่อชั่วโมงการทำงาน
   * **คณะกรรมการแท็กซี่และลีมูซีนแห่งนครนิวยอร์ก (NYC TLC & City Planners):** วางแผนนโยบายการจราจร กำหนดจุดจอดรถแท็กซี่สาธารณะ (Taxi Stands) และประเมินความต้องการโครงสร้างพื้นฐานด้านการคมนาคมที่เชื่อมโยงกับสนามบิน JFK และ LaGuardia

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**

* **ประเภทงาน (Task Type):** การถดถอยเชิงพื้นที่และเวลา (Spatio-Temporal Regression Task)
* **ตัวแปรเป้าหมาย (Target Variable):** `demand_count` คือ จำนวนการเดินทางด้วยรถแท็กซี่ที่เริ่มต้นขึ้นจริงในโซน `PULocationID` ในช่วงชั่วโมง `pickup_hour` นั้นๆ มีค่าเป็นจำนวนเต็มตั้งแต่ 0 ขึ้นไป ($y \in \mathbb{Z}_{\ge 0}$)
* **คุณลักษณะพิเศษของปัญหา:** ข้อมูลมีการแจกแจงแบบเบ้ขวาจัด (Heavily Right-skewed) และมีภาวะศูนย์จำลองจำนวนมาก (Zero-Inflated) ในช่วงเวลาดึกหรือโซนรอบนอก ซึ่งต้องใช้โมเดลกลุ่ม Tree-based Gradient Boosting และการเติมเต็มตารางกริดแบบเต็มรูปแบบ (Full Cartesian Grid) เพื่อป้องกันความลำเอียงของโมเดล

### หน่วยของการวิเคราะห์ (Unit of Analysis)
* **ก่อนยุบรวม (Raw Trips):** 1 แถว แทน **การเดินทางด้วยรถแท็กซี่ 1 เที่ยว (Single Taxi Trip)** โดยมีรายละเอียดจุดเริ่มต้น จุดสิ้นสุด ค่าโดยสาร และระยะทาง
* **หลังยุบรวมเพื่อสร้างแบบจำลอง (Aggregated Spatio-Temporal Model):** 1 แถว แทน **คู่ของโซนรับผู้โดยสารและช่วงเวลา 1 ชั่วโมง (`PULocationID` + `pickup_hour`)** ซึ่งประกอบด้วยตารางกริดสมบูรณ์ทุกชั่วโมงและทุกโซน (รวมถึงชั่วโมงที่มีการเรียกรถ 0 ครั้ง) เพื่อสะท้อนความต้องการรถในมิติพื้นที่และเวลาอย่างแท้จริง

## 3. Data Loading
ในขั้นตอนนี้ เราทำการดึงข้อมูลจริงของ Yellow Taxi จากหน้าเว็บทางการของ NYC TLC ([TLC Trip Record Data](https://www.nyc.gov/site/tlc/about/tlc-trip-record-data.page)) ครบทั้ง 4 ฤดูกาล (ปี 2024) พร้อมกับตาราง `taxi_zone_lookup.csv` เชื่อมโยงข้อมูล (Merge) ด้วยคีย์ `PULocationID = LocationID` เพื่อดึงข้อมูล `Borough` และ `service_zone`

```python
# Load data & merge with taxi zone lookup
# ข้อมูลจากหน้าทางการของ NYC TLC: https://www.nyc.gov/site/tlc/about/tlc-trip-record-data.page
import os
import sys
import time
import urllib.request
import numpy as np
import pandas as pd

# ตรวจสอบและติดตั้งแพ็กเกจที่จำเป็นอัตโนมัติหากรันในสภาพแวดล้อมใหม่ (เช่น Google Colab)
for pkg in ['pyarrow', 'lightgbm']:
    try:
        __import__(pkg)
    except ImportError:
        import subprocess
        print(f"กำลังติดตั้ง {pkg}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])

import pyarrow.parquet as pq

# สร้างโฟลเดอร์ data หากยังไม่มี
data_dir = "data"
os.makedirs(data_dir, exist_ok=True)

# ฟังก์ชันช่วยดาวน์โหลดไฟล์แบบ Chunk พร้อม User-Agent เพื่อความเสถียรสูงสุดบน Google Colab
def robust_download(url, target_path, min_mb=0.01):
    need_dl = True
    if os.path.exists(target_path):
        cur_mb = os.path.getsize(target_path) / (1024 * 1024)
        if cur_mb >= min_mb:
            need_dl = False
        else:
            print(f"ไฟล์เดิมไม่สมบูรณ์ ({cur_mb:.2f} MB < {min_mb} MB) กำลังดาวน์โหลดใหม่...")
            try:
                os.remove(target_path)
            except OSError:
                pass
    if need_dl:
        print(f"กำลังดาวน์โหลด {os.path.basename(target_path)} จาก {url}...")
        t_dl = time.time()
        req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'})
        with urllib.request.urlopen(req) as resp, open(target_path, 'wb') as out_f:
            while True:
                chunk = resp.read(1024 * 1024)
                if not chunk:
                    break
                out_f.write(chunk)
        size_mb = os.path.getsize(target_path) / (1024 * 1024)
        print(f"ดาวน์โหลดสำเร็จ: {size_mb:.2f} MB ใน {time.time()-t_dl:.1f} วินาที")

# 1. จัดการดาวน์โหลดและโหลดตารางโซนพื้นที่ taxi_zone_lookup.csv
zone_lookup_url = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"
zone_lookup_path = os.path.join(data_dir, "taxi_zone_lookup.csv")
robust_download(zone_lookup_url, zone_lookup_path, min_mb=0.01)

zone_df = pd.read_csv(zone_lookup_path)
print(f"Loaded taxi_zone_lookup.csv: {len(zone_df):,} zones")
print(zone_df.head(3))

# 2. นิยามไฟล์ข้อมูลจริงครบทั้ง 4 ฤดูกาล (2024) พร้อม URL ทางการสำหรับดาวน์โหลดอัตโนมัติ
season_files = [
    {
        "season_name": "Winter (ม.ค. 2024)", "season_code": 1, "month": 1, "year": 2024, "days": 31,
        "file": os.path.join(data_dir, "yellow_tripdata_2024-01.parquet"),
        "url": "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-01.parquet"
    },
    {
        "season_name": "Spring (เม.ย. 2024)", "season_code": 2, "month": 4, "year": 2024, "days": 30,
        "file": os.path.join(data_dir, "yellow_tripdata_2024-04.parquet"),
        "url": "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-04.parquet"
    },
    {
        "season_name": "Summer (ก.ค. 2024)", "season_code": 3, "month": 7, "year": 2024, "days": 31,
        "file": os.path.join(data_dir, "yellow_tripdata_2024-07.parquet"),
        "url": "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-07.parquet"
    },
    {
        "season_name": "Autumn (ต.ค. 2024)", "season_code": 4, "month": 10, "year": 2024, "days": 31,
        "file": os.path.join(data_dir, "yellow_tripdata_2024-10.parquet"),
        "url": "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-10.parquet"
    },
]

use_cols = [
    'tpep_pickup_datetime', 'tpep_dropoff_datetime',
    'PULocationID', 'DOLocationID',
    'passenger_count', 'trip_distance',
    'fare_amount', 'total_amount'
]

raw_row_counts = {}
loaded_dfs = []
total_raw_rows = 0

print("\n--- เริ่มโหลดข้อมูลดิบครบ 4 ฤดูกาล (Memory-Safe Downcasting & Colab Auto-Download) ---")
t_load_start = time.time()

for item in season_files:
    # ดาวน์โหลดไฟล์โดยอัตโนมัติหากยังไม่มีในเครื่อง (พร้อมตรวจสอบความสมบูรณ์ขั้นต่ำ 30 MB)
    robust_download(item['url'], item['file'], min_mb=30.0)

    meta = pq.read_metadata(item['file'])
    n_rows = meta.num_rows
    raw_row_counts[item['season_name']] = n_rows
    total_raw_rows += n_rows
    print(f"กำลังอ่าน {item['season_name']}: {n_rows:,} แถว, {meta.num_columns} คอลัมน์")

    # โหลดเฉพาะคอลัมน์สำคัญและแปลงชนิดข้อมูลเพื่อประหยัด RAM
    df_chunk = pd.read_parquet(item['file'], columns=use_cols)
    df_chunk['PULocationID'] = df_chunk['PULocationID'].astype(np.int16)
    df_chunk['DOLocationID'] = df_chunk['DOLocationID'].astype(np.int16)
    df_chunk['passenger_count'] = df_chunk['passenger_count'].astype(np.float32)
    df_chunk['trip_distance'] = df_chunk['trip_distance'].astype(np.float32)
    df_chunk['fare_amount'] = df_chunk['fare_amount'].astype(np.float32)
    df_chunk['total_amount'] = df_chunk['total_amount'].astype(np.float32)
    df_chunk['season'] = np.int8(item['season_code'])
    loaded_dfs.append(df_chunk)

raw_df = pd.concat(loaded_dfs, ignore_index=True)
del loaded_dfs
load_time = time.time() - t_load_start

print(f"\nโหลดข้อมูลเสร็จสิ้นใน {load_time:.2f} วินาที")
print(f"จำนวนแถวดิบรวมทั้ง 4 ฤดูกาล: {total_raw_rows:,} แถว (สูงกว่าเกณฑ์ขั้นต่ำ 1,000,000 แถว)")
print(f"ขนาดหน่วยความจำของข้อมูลดิบใน RAM: {raw_df.memory_usage(deep=True).sum() / (1024**2):.2f} MB")

# 3. ตรวจสอบการเชื่อมข้อมูล (Merge) กับตารางโซน taxi_zone_lookup
rows_before_merge = len(raw_df)
raw_merged_df = raw_df.merge(zone_df[['LocationID', 'Borough', 'service_zone']],
                             left_on='PULocationID', right_on='LocationID', how='left')
rows_after_merge = len(raw_merged_df)

print(f"\n--- ผลการเชื่อมข้อมูล (Merge) ด้วยคีย์ PULocationID = LocationID ---")
print(f"จำนวนแถวก่อนการ Merge: {rows_before_merge:,} แถว")
print(f"จำนวนแถวหลังการ Merge:  {rows_after_merge:,} แถว")
assert rows_before_merge == rows_after_merge, "จำนวนแถวไม่ตรงกันหลังการ Merge!"
print("การเชื่อมโยงข้อมูลสมบูรณ์แบบ แถวคงอยู่ครบ 100%")
```

## 4. Data Quality Assessment
ทำการตรวจประเมินคุณภาพของข้อมูลดิบอย่างละเอียดและรายงานเป็นตัวเลขสถิติและร้อยละใน 5 มิติหลัก:
1. **ชนิดข้อมูล (Data Types) และปริมาณการใช้หน่วยความจำ (Memory Usage)**
2. **ค่าสูญหาย (Missing Values):** จำนวนและร้อยละต่อคอลัมน์
3. **ข้อมูลซ้ำซ้อน (Duplicate Rows):** ตรวจสอบแถวที่ซ้ำกันทุกคอลัมน์
4. **ค่าผิดปกติหรือเป็นไปไม่ได้เชิงตรรกะ (Invalid / Impossible Values):**
   * วันเวลาที่รับผู้โดยสารอยู่นอกช่วงของแต่ละฤดูกาล (Timestamp Out-of-Range)
   * เวลาส่งผู้โดยสารเกิดก่อนหรือพร้อมกับเวลารับผู้โดยสาร (`tpep_dropoff_datetime <= tpep_pickup_datetime`) หรือระยะเวลาเดินทางนานเกิน 24 ชั่วโมง
   * ระยะทางติดลบหรือเป็นศูนย์ (`trip_distance <= 0`) หรือระยะทางไกลเกินจริงผิดปกติ (`trip_distance > 200` ไมล์)
   * ค่าโดยสารติดลบหรือเป็นศูนย์ (`fare_amount <= 0` หรือ `total_amount <= 0`)
   * จำนวนผู้โดยสารเป็นค่าว่าง ศูนย์ หรือมากกว่า 9 คน
5. **ความไม่สอดคล้องของรหัสพื้นที่ (Invalid Location IDs):** รหัสโซนที่ไม่ทราบค่า (Unknown Zone 264, 265) หรือไม่อยู่ในฐานข้อมูลโซนของ TLC

```python
# Data Quality Assessment across 13.38M raw rows
print("=== 1. Data Types and Memory Usage ===")
mem_summary = pd.DataFrame({
    'Data Type': raw_df.dtypes,
    'Memory (MB)': (raw_df.memory_usage(deep=True) / (1024**2)).round(2)
})
print(mem_summary)

print("\n=== 2. Missing Values Analysis ===")
missing_count = raw_df.isnull().sum()
missing_pct = (missing_count / len(raw_df) * 100).round(4)
missing_table = pd.DataFrame({
    'Missing Rows': missing_count,
    'Missing Percentage (%)': missing_pct
})
print(missing_table[missing_table['Missing Rows'] > 0])
print(f"คอลัมน์ที่มีค่าสูญหายมีเพียงคอลัมน์เดียวคือ passenger_count จำนวน {missing_count['passenger_count']:,} แถว ({missing_pct['passenger_count']}%)")

print("\n=== 3. Duplicate Rows Analysis ===")
dup_rows = raw_df.duplicated().sum()
print(f"จำนวนแถวที่ซ้ำกันทุกคอลัมน์ (Exact Duplicates): {dup_rows:,} แถว ({dup_rows / len(raw_df) * 100:.6f}%)")

print("\n=== 4. Invalid / Impossible Values Assessment ===")
# 4.1 Timestamp Out-of-Range per season
season_boundaries = {
    1: (pd.Timestamp('2024-01-01 00:00:00'), pd.Timestamp('2024-01-31 23:59:59')),
    2: (pd.Timestamp('2024-04-01 00:00:00'), pd.Timestamp('2024-04-30 23:59:59')),
    3: (pd.Timestamp('2024-07-01 00:00:00'), pd.Timestamp('2024-07-31 23:59:59')),
    4: (pd.Timestamp('2024-10-01 00:00:00'), pd.Timestamp('2024-10-31 23:59:59')),
}
out_of_range_mask = pd.Series(False, index=raw_df.index)
for s_code, (start_dt, end_dt) in season_boundaries.items():
    s_idx = (raw_df['season'] == s_code)
    out_of_range_mask |= (s_idx & ((raw_df['tpep_pickup_datetime'] < start_dt) | (raw_df['tpep_pickup_datetime'] > end_dt)))

out_range_count = out_of_range_mask.sum()
print(f"- เวลาเดินทางอยู่นอกช่วงฤดูกาลที่ศึกษา (Out-of-Range): {out_range_count:,} แถว ({out_range_count/len(raw_df)*100:.4f}%)")

# 4.2 Dropoff before/equal pickup or duration > 24 hours
duration_minutes = (raw_df['tpep_dropoff_datetime'] - raw_df['tpep_pickup_datetime']).dt.total_seconds() / 60.0
invalid_duration = (duration_minutes < 1.0) | (duration_minutes > 1440.0)
print(f"- ระยะเวลาเดินทางผิดปกติ (< 1 นาที หรือ > 24 ชม.): {invalid_duration.sum():,} แถว ({invalid_duration.sum()/len(raw_df)*100:.4f}%)")

# 4.3 Trip distance <= 0 or > 200 miles
invalid_dist = (raw_df['trip_distance'] <= 0) | (raw_df['trip_distance'] > 200)
print(f"- ระยะทางเป็นศูนย์ ติดลบ หรือเกิน 200 ไมล์: {invalid_dist.sum():,} แถว ({invalid_dist.sum()/len(raw_df)*100:.4f}%)")

# 4.4 Non-positive fares
invalid_fare = (raw_df['fare_amount'] <= 0) | (raw_df['total_amount'] <= 0)
print(f"- ค่าโดยสารหรือยอดรวมติดลบหรือเป็นศูนย์: {invalid_fare.sum():,} แถว ({invalid_fare.sum()/len(raw_df)*100:.4f}%)")

# 4.5 Invalid passenger counts
invalid_pass = (raw_df['passenger_count'].isna()) | (raw_df['passenger_count'] <= 0) | (raw_df['passenger_count'] > 9)
print(f"- จำนวนผู้โดยสารสูญหาย เป็นศูนย์ หรือมากกว่า 9 คน: {invalid_pass.sum():,} แถว ({invalid_pass.sum()/len(raw_df)*100:.4f}%)")

# 4.6 Unknown / Out-of-bounds TLC Location IDs (264=Unknown, 265=NA)
valid_locs = set(zone_df[~zone_df['LocationID'].isin([264, 265])]['LocationID'])
invalid_loc = (~raw_df['PULocationID'].isin(valid_locs)) | (~raw_df['DOLocationID'].isin(valid_locs))
print(f"- รหัสโซนที่ไม่ทราบค่า (Unknown Zones 264/265): {invalid_loc.sum():,} แถว ({invalid_loc.sum()/len(raw_df)*100:.4f}%)")
```

### ขั้นตอนสำคัญหลังทำความสะอาด: การสร้าง Full Cartesian Grid และแก้ปัญหา Zero-Demand Trap
หากเรายุบรวมข้อมูลเฉพาะชั่วโมงที่มีคนเรียกรถ จะได้ข้อมูลเพียง **285,161 แถว** ซึ่งละเลยช่วงเวลาดึกหรือพื้นที่รอบนอกที่ไม่มีคนเรียกรถ ทำให้โมเดลเรียนรู้เฉพาะสภาวะที่มีคน และจะทำนายเกินจริง (Overestimation Bias) อย่างรุนแรงในช่วงที่ไม่มีความต้องการ

เราจึงสร้าง **Full Cartesian Grid** เต็มรูปแบบ:
$$	ext{Full Grid} = 263 	ext{ โซนที่ใช้งานจริง} 	imes 2,952 	ext{ ชั่วโมง (ครอบคลุมทั้ง 4 ฤดูกาล)} = \mathbf{776,376} 	ext{ แถว}$$
และทำการเติมค่าศูนย์ (`demand_count = 0`) ในชั่วโมงที่ไม่มีการเรียกรถ (จำนวน 491,215 แถว หรือ 63.27% ของตารางกริดทั้งหมด) ส่งผลให้ชุดข้อมูลมีความสมบูรณ์เชิงสถิติและสะท้อนความจริงอย่างแท้จริง

```python
# Data Cleaning Pipeline & Full Cartesian Grid Generation
t_clean_start = time.time()

# 1. รหัสโซนที่ถูกต้อง (ตัด 264 Unknown และ 265 N/A)
valid_zone_ids = zone_df[~zone_df['LocationID'].isin([264, 265])]['LocationID'].values

# ดำเนินการทำความสะอาดข้อมูลตามลำดับ
# ขั้นที่ 1: ตรวจสอบวันเวลา (คำนวณ mask หากยังไม่มี)
if 'out_of_range_mask' not in globals():
    season_boundaries = {
        1: (pd.Timestamp('2024-01-01 00:00:00'), pd.Timestamp('2024-01-31 23:59:59')),
        2: (pd.Timestamp('2024-04-01 00:00:00'), pd.Timestamp('2024-04-30 23:59:59')),
        3: (pd.Timestamp('2024-07-01 00:00:00'), pd.Timestamp('2024-07-31 23:59:59')),
        4: (pd.Timestamp('2024-10-01 00:00:00'), pd.Timestamp('2024-10-31 23:59:59')),
    }
    out_of_range_mask = pd.Series(False, index=raw_df.index)
    for s_code, (start_dt, end_dt) in season_boundaries.items():
        s_idx = (raw_df['season'] == s_code)
        out_of_range_mask |= (s_idx & ((raw_df['tpep_pickup_datetime'] < start_dt) | (raw_df['tpep_pickup_datetime'] > end_dt)))

clean_mask_1 = ~out_of_range_mask
df_step1 = raw_df[clean_mask_1]

# ขั้นที่ 2: ตรวจสอบระยะเวลาเดินทาง (1 - 1440 นาที)
dur_min = (df_step1['tpep_dropoff_datetime'] - df_step1['tpep_pickup_datetime']).dt.total_seconds() / 60.0
clean_mask_2 = (dur_min >= 1.0) & (dur_min <= 1440.0)
df_step2 = df_step1[clean_mask_2]

# ขั้นที่ 3: ตรวจสอบระยะทาง (0 < dist <= 200)
clean_mask_3 = (df_step2['trip_distance'] > 0) & (df_step2['trip_distance'] <= 200)
df_step3 = df_step2[clean_mask_3]

# ขั้นที่ 4: ตรวจสอบค่าโดยสาร (> 0)
clean_mask_4 = (df_step3['fare_amount'] > 0) & (df_step3['total_amount'] > 0)
df_step4 = df_step3[clean_mask_4]

# ขั้นที่ 5: ตรวจสอบผู้โดยสาร (1 - 9 คน)
clean_mask_5 = (df_step4['passenger_count'].notna()) & (df_step4['passenger_count'] >= 1) & (df_step4['passenger_count'] <= 9)
df_step5 = df_step4[clean_mask_5]

# ขั้นที่ 6: ตรวจสอบโซนพื้นที่
clean_mask_6 = df_step5['PULocationID'].isin(valid_zone_ids)
df_step6 = df_step5[clean_mask_6]

# ขั้นที่ 7: ลบข้อมูลซ้ำ
cleaned_trips_df = df_step6.drop_duplicates().copy()

print(f"=== ผลการทำความสะอาดข้อมูลการเดินทาง (Cleaned Trips) ===")
print(f"จำนวนแถวเริ่มต้น: {len(raw_df):,} แถว")
print(f"จำนวนแถวคงเหลือ:  {len(cleaned_trips_df):,} แถว (คิดเป็น {len(cleaned_trips_df)/len(raw_df)*100:.2f}%)")
print(f"เวลาที่ใช้ในการทำความสะอาด: {time.time()-t_clean_start:.2f} วินาที")

# เคลียร์หน่วยความจำของดาต้าเฟรมขั้นกลาง
del df_step1, df_step2, df_step3, df_step4, df_step5, df_step6

# ขั้นที่ 8: ยุบรวมข้อมูลเป็นระดับ PULocationID x pickup_hour และสร้าง Full Cartesian Grid
cleaned_trips_df['pickup_hour'] = cleaned_trips_df['tpep_pickup_datetime'].dt.floor('h')
agg_demand = cleaned_trips_df.groupby(['PULocationID', 'pickup_hour']).size().reset_index(name='demand_count')

# สร้าง Cartesian Grid เต็มรูปแบบสำหรับแต่ละฤดูกาล
grid_season_dfs = []
for item in season_files:
    hours = pd.date_range(start=f"{item['year']}-{item['month']:02d}-01 00:00:00",
                          end=f"{item['year']}-{item['month']:02d}-{item['days']:02d} 23:00:00",
                          freq='h')
    idx = pd.MultiIndex.from_product([valid_zone_ids, hours], names=['PULocationID', 'pickup_hour'])
    g_df = pd.DataFrame(index=idx).reset_index()
    g_df['season'] = np.int8(item['season_code'])
    grid_season_dfs.append(g_df)

full_grid = pd.concat(grid_season_dfs, ignore_index=True)
del grid_season_dfs

# เชื่อมโยงข้อมูล demand เข้ากับตารางกริดและเติม 0
full_grid = pd.merge(full_grid, agg_demand, on=['PULocationID', 'pickup_hour'], how='left')
full_grid['demand_count'] = full_grid['demand_count'].fillna(0).astype(np.int32)

# เชื่อมโยง Borough และ service_zone จาก lookup
valid_zone_lookup = zone_df[zone_df['LocationID'].isin(valid_zone_ids)][['LocationID', 'Borough', 'service_zone']].copy()
full_grid = full_grid.merge(valid_zone_lookup, left_on='PULocationID', right_on='LocationID', how='left')
full_grid.drop(columns=['LocationID'], inplace=True)

print(f"\n=== ขนาดของ Full Cartesian Grid ครบ 4 ฤดูกาล ===")
print(f"จำนวนแถวทั้งหมดในตารางกริด: {len(full_grid):,} แถว")
print(f"จำนวนช่วงเวลาที่มีการเรียกรถ (>0): {len(agg_demand):,} แถว ({len(agg_demand)/len(full_grid)*100:.2f}%)")
print(f"จำนวนช่วงเวลาที่ไม่มีการเรียกรถ (=0): {(full_grid['demand_count']==0).sum():,} แถว ({(full_grid['demand_count']==0).sum()/len(full_grid)*100:.2f}%)")
print(f"ตัวอย่างตารางกริด 5 แถวแรก:")
print(full_grid.head(5))
```

**สิ่งที่ค้นพบจาก EDA และผลต่อการเตรียมข้อมูล/การเลือกโมเดล**

จากการสำรวจข้อมูลเชิงลึก เราค้นพบประเด็นสำคัญ 4 ประการที่ส่งผลโดยตรงต่อการออกแบบระบบและการสร้างแบบจำลอง:

1. **การแจกแจงของ Target มีความเบ้ขวาจัด (Skewness = 8.79) และมีภาวะศูนย์สูง (Zero-Inflation):**
   * *สิ่งที่ค้นพบ:* ค่าเฉลี่ยของ `demand_count` อยู่ที่ 14.97 เที่ยว/ชม./โซน แต่มัธยฐานอยู่ที่เพียง 0 เที่ยว/ชม./โซน และกว่า 63% ของตารางกริดมีค่าเป็น 0 ขณะที่ช่วงพีคในโซนสำคัญสูงถึงกว่า 900 เที่ยว/ชม.
   * *ผลต่อการเตรียมข้อมูลและการเลือกโมเดล:* บังคับให้ต้องสร้าง **Full Cartesian Grid** เติมค่าศูนย์ในชั่วโมงที่ไม่มีการเรียกรถเพื่อป้องกัน Overestimation Bias และเลือกใช้ **โมเดลกลุ่ม Tree-based (LightGBM)** ที่ทนทานต่อการแจกแจงแบบเบ้ขวาและจัดการกับ Zero-Inflation ได้ดีเยี่ยมโดยไม่ต้องแปลงค่า Target

2. **วัฏจักรเวลาที่เด่นชัดในรอบ 24 ชั่วโมงและ 168 ชั่วโมง (Daily & Weekly Seasonality):**
   * *สิ่งที่ค้นพบ:* Heatmap และ Autocorrelation แสดงให้เห็นว่าความต้องการมีสหสัมพันธ์สูงมากกับชั่วโมงเดียวกันของสัปดาห์ก่อน (`lag_168h` มีค่า $r = 0.942$) และชั่วโมงเดียวกันของวันก่อนหน้า (`lag_24h` มีค่า $r = 0.938$) รวมทั้งช่วงเวลาเร่งด่วนเช้า (8:00–9:00) และเย็น (17:00–19:00) แตกต่างจากช่วงวันหยุดอย่างสิ้นเชิง
   * *ผลต่อการเตรียมข้อมูล:*
   นำไปสู่การออกแบบ **Feature Engineering (พาร์ท 7)** เพื่อสร้าง Lag Features (`lag_1h`, `lag_24h`, `lag_168h`) และแปลงวัฏจักรเวลาให้อยู่ในรูปต่อเนื่องผ่าน **Cyclical Encoding** (`hour_sin/cos`, `dow_sin/cos`)
   
3. **ผลกระทบของฤดูกาลต่อพฤติกรรมการเรียกรถ (Seasonal Shift):**
   * *สิ่งที่ค้นพบ:* ในฤดูใบไม้ร่วง (ตุลาคม) มีปริมาณการเดินทางรวมสูงสุด (กว่า 3.28 ล้านเที่ยวที่ผ่านการคลีน) เนื่องจากเป็นช่วงเปิดเทอมและกิจกรรมทางธุรกิจ ขณะที่ฤดูหนาว (มกราคม) มีปริมาณการเรียกรถต่ำกว่า โดยมีรูปแบบการเดินทางช่วงกลางคืนที่ลดลงอย่างมีนัยสำคัญ
   * *ผลต่อการเตรียมข้อมูล:* ต้องใส่ตัวแปร `season` (One-Hot Encoded) และ `month_sin/cos` เพื่อให้โมเดลสามารถเรียนรู้พลวัตและการเปลี่ยนแปลงพฤติกรรมตลอดทั้ง 4 ฤดูกาลได้แม่นยำ

4. **ความกระจุกตัวเชิงพื้นที่สูงมากในเขต Manhattan และสนามบิน (Spatial Concentration):**
   * *สิ่งที่ค้นพบ:* เกินกว่า 90% ของความต้องการกระจุกตัวอยู่ในแมนฮัตตันและสองสนามบินหลัก (JFK และ LaGuardia ในเขตควีนส์) โดยโซนรอบนอก (Bronx, Staten Island) มีความต้องการเฉลี่ยต่ำกว่า 1 เที่ยว/ชม.
   * *ผลต่อการเตรียมข้อมูลและการประเมิน:* ต้องนำตัวแปร `Borough` และ `service_zone` มาร่วมในโมเดล และเป็นเหตุผลสำคัญที่ทำให้ต้องทำ **Borough-level Error Analysis (พาร์ท 11)** เพื่อตรวจสอบความแม่นยำรายเขตอย่างเที่ยงธรรมโดยไม่ให้ความหนาแน่นในแมนฮัตตันมาบดบังข้อผิดพลาดในโซนรอบนอก

## 7. Feature Engineering
จากผลการสำรวจข้อมูลเชิงลึกในขั้นตอน EDA ที่ชี้ให้เห็นถึงความสำคัญของมิติเวลา วัฏจักรรายสัปดาห์/รายวัน ฤดูกาล และความกระจุกตัวเชิงพื้นที่ เราจึงได้ออกแบบและสร้างตัวแปรขึ้นมาทั้งหมด 25 ตัวแปร เพื่อแปลงข้อมูลดิบให้สะท้อนพฤติกรรมจริงและช่วยให้โมเดล Machine Learning สามารถเรียนรู้ได้อย่างมีประสิทธิภาพสูงสุด ครอบคลุม 5 มิติหลัก ดังตารางสรุปด้านล่างนี้:

### ตารางสรุปคุณลักษณะตัวแปร (Feature Engineering Table)

| กลุ่มตัวแปร | ชื่อตัวแปร (Feature) | สร้างจาก | เหตุผลทางธุรกิจและสถิติ |
|---|---|---|---|
| **มิติเวลาและปฏิทิน** | `hour`, `day_of_week`, `day_of_month`, `month` | แยกจาก `pickup_hour` | ระบุตำแหน่งเวลาในรอบวัน รอบสัปดาห์ และรอบปี |
| **ตัวบ่งชี้พฤติกรรม** | `is_weekend` | `day_of_week >= 5` | แยกพฤติกรรมการเดินทางวันหยุดออกจากวันทำงาน |
| | `is_rush_hour` | ชั่วโมง 7–9 และ 16–19 | จับช่วงเวลาที่มีความต้องการหนาแน่นสูงเป็นพิเศษ (อ้างอิงจาก Heatmap ใน EDA) |
| | `is_us_holiday` | วันหยุดราชการสหรัฐฯ ในปี 2024 | จับพฤติกรรมวันหยุดนักขัตฤกษ์ที่คล้ายวันหยุดสุดสัปดาห์ |
| **วัฏจักรตรีโกณมิติ** | `hour_sin`, `hour_cos` | $\sin/\cos(2\pi 	imes 	ext{hour} / 24)$ | ทำให้โมเดลเรียนรู้ว่าชั่วโมง 23:00 และ 00:00 อยู่ติดกัน |
| | `dow_sin`, `dow_cos` | $\sin/\cos(2\pi 	imes 	ext{dow} / 7)$ | เชื่อมโยงวันอาทิตย์ (6) เข้ากับวันจันทร์ (0) อย่างต่อเนื่อง |
| | `month_sin`, `month_cos` | $\sin/\cos(2\pi 	imes 	ext{month} / 12)$ | เชื่อมโยงรอบฤดูกาลตลอดทั้ง 12 เดือน |
| **ตัวแปรย้อนหลัง (Lags)** | `lag_1h`, `lag_2h`, `lag_3h` | `.shift(1, 2, 3)` ตามโซน | โมเมนตัมความต้องการในระยะสั้น (Short-term Autocorrelation) |
| | `lag_24h` | `.shift(24)` ตามโซน | พฤติกรรมความต้องการ ณ เวลาเดียวกันของเมื่อวาน |
| | `lag_168h` | `.shift(168)` ตามโซน | พฤติกรรมความต้องการ ณ เวลาเดียวกันของสัปดาห์ก่อน ($r = 0.942$ จาก EDA) |
| **ค่าสถิติเคลื่อนที่ (Rolling)** | `rolling_mean_3h`, `6h`, `24h` | `.shift(1).rolling(w).mean()` | ค่าเฉลี่ยแนวโน้มระดับสั้น กลาง และยาว โดยปราศจาก Data Leakage |
| | `rolling_std_24h` | `.shift(1).rolling(24).std()` | สะท้อนความผันผวนของความต้องการในรอบ 24 ชั่วโมง |
| **มิติเชิงพื้นที่และฤดูกาล** | `Borough`, `service_zone` | ตารางอ้างอิงโซน TLC | ควบคุมอิทธิพลของระดับความเจริญและข้อกำหนดการเดินรถ |
| | `season` | ฤดูกาล (1=Winter, 2=Spring, 3=Summer, 4=Autumn) | จับพฤติกรรมสภาพอากาศและการท่องเที่ยวตลอด 4 ฤดูกาล |

### การยืนยันและพิสูจน์การป้องกันข้อมูลรั่วไหล (Strict Data Leakage Prevention Proof)
เพื่อความโปร่งใสทางวิชาการและรับประกันว่าระบบสามารถนำไปใช้จริงในหน้างานได้ ไปป์ไลน์การสร้างฟีเจอร์จึงยึดมั่นใน 3 กฎเหล็กอย่างเคร่งครัด:

1. **การคำนวณ Rolling Statistics และ Lags ทั้งหมด:** ต้องเรียกใช้คำสั่ง `.shift(1)` เสมอก่อนนำไปเข้าฟังก์ชัน `.rolling()` เพื่อรับประกันว่าค่าสถิติที่ได้จะใช้เฉพาะข้อมูลในอดีตเท่านั้น และไม่รวมข้อมูลของชั่วโมงเป้าหมายที่กำลังจะพยากรณ์
2. **ไม่ใช้ตัวแปรที่รู้หลังการเดินทางเสร็จสิ้น:** เราไม่นำตัวแปร `dropoff_datetime`, `trip_distance`, `fare_amount`, หรือ `total_amount` ของชั่วโมงเป้าหมายมาใช้เป็นฟีเจอร์เด็ดขาด เพราะตัวแปรเหล่านี้จะทราบค่าได้ก็ต่อเมื่อการเดินทางสิ้นสุดลงแล้วเท่านั้น
3. **การแยกคำนวณตามโซนและฤดูกาล:** การ `.shift()` ถูกจัดกลุ่มแยกตาม `['PULocationID', 'season']` เพื่อป้องกันไม่ให้ข้อมูลของฤดูกาลถัดไปรั่วไหลข้ามมายังฤดูกาลก่อนหน้า

```python
# Feature Engineering Implementation
t_feat_start = time.time()

# 1. ปฏิทินวันหยุดสำคัญของสหรัฐฯ ในปี 2024
us_holidays_2024 = {
    pd.Timestamp('2024-01-01').date(), # New Year's Day
    pd.Timestamp('2024-01-15').date(), # Martin Luther King Jr. Day
    pd.Timestamp('2024-07-04').date(), # Independence Day
    pd.Timestamp('2024-10-14').date()  # Columbus / Indigenous Peoples Day
}

# 2. ตัวแปรเวลาและปฏิทิน
full_grid['day_of_month'] = full_grid['pickup_hour'].dt.day.astype(np.int8)
full_grid['month'] = full_grid['pickup_hour'].dt.month.astype(np.int8)
full_grid['is_weekend'] = full_grid['day_of_week'].isin([5, 6]).astype(np.int8)
full_grid['is_rush_hour'] = full_grid['hour'].isin([7, 8, 9, 16, 17, 18, 19]).astype(np.int8)
full_grid['is_us_holiday'] = full_grid['pickup_hour'].dt.date.isin(us_holidays_2024).astype(np.int8)

# 3. Cyclical Encoding
full_grid['hour_sin'] = np.sin(2 * np.pi * full_grid['hour'] / 24.0).astype(np.float32)
full_grid['hour_cos'] = np.cos(2 * np.pi * full_grid['hour'] / 24.0).astype(np.float32)
full_grid['dow_sin'] = np.sin(2 * np.pi * full_grid['day_of_week'] / 7.0).astype(np.float32)
full_grid['dow_cos'] = np.cos(2 * np.pi * full_grid['day_of_week'] / 7.0).astype(np.float32)
full_grid['month_sin'] = np.sin(2 * np.pi * full_grid['month'] / 12.0).astype(np.float32)
full_grid['month_cos'] = np.cos(2 * np.pi * full_grid['month'] / 12.0).astype(np.float32)

# 4. Lag Features และ Rolling Statistics (จัดกลุ่มตามโซนและฤดูกาล เรียงตามลำดับเวลาอย่างเข้มงวด)
full_grid.sort_values(by=['PULocationID', 'pickup_hour'], inplace=True)

# Lags
for lag in [1, 2, 3, 24, 168]:
    full_grid[f'lag_{lag}h'] = full_grid.groupby(['PULocationID', 'season'])['demand_count'].shift(lag).astype(np.float32)

# Rolling Statistics (คำนวณหลังจาก shift(1) เพื่อป้องกัน Data Leakage)
for w in [3, 6, 24]:
    full_grid[f'rolling_mean_{w}h'] = full_grid.groupby(['PULocationID', 'season'])['demand_count'].transform(
        lambda s: s.shift(1).rolling(w).mean()
    ).astype(np.float32)

full_grid['rolling_std_24h'] = full_grid.groupby(['PULocationID', 'season'])['demand_count'].transform(
    lambda s: s.shift(1).rolling(24).std()
).fillna(0).astype(np.float32)

print(f"Feature engineering เสร็จสิ้นใน {time.time()-t_feat_start:.2f} วินาที")
print(f"มิติของข้อมูลหลังสร้าง Features: {full_grid.shape}")

# 5. ตัดช่วง Warm-up 7 วันแรกของแต่ละฤดูกาล (เนื่องจาก lag_168h ต้องใช้ข้อมูลย้อนหลัง 7 วัน)
clean_grid = full_grid.dropna().copy()
print(f"จำนวนแถวหลังตัดช่วง Warm-up (พร้อมสำหรับแบบจำลอง): {len(clean_grid):,} แถว")
```
