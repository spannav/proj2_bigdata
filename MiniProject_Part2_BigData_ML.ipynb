> *"เราสามารถพัฒนาระบบการเรียนรู้ของเครื่อง (Machine Learning Pipeline) เพื่อพยากรณ์จำนวนความต้องการเรียกรถแท็กซี่ (`demand_count`) ล่วงหน้า 1 ชั่วโมงในแต่ละโซนพื้นที่รับผู้โดยสาร (`PULocationID`) ได้แม่นยำเพียงใดตลอดทั้ง 4 ฤดูกาลของปี และปัจจัยด้านเวลา วัฏจักรย้อนหลัง ฤดูกาล หรือพื้นที่ มีอิทธิพลต่อความแม่นยำในการพยากรณ์มากที่สุด?"*

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**

* **ประเภทงาน (Task Type):** การถดถอยเชิงพื้นที่และเวลา (Spatio-Temporal Regression Task)

* **คุณลักษณะพิเศษของปัญหา:** ข้อมูลมีการแจกแจงแบบเบ้ขวาจัด (Heavily Right-skewed) และมีภาวะศูนย์จำลองจำนวนมาก (Zero-Inflated) ในช่วงเวลาดึกหรือโซนรอบนอก ซึ่งต้องใช้โมเดลกลุ่ม Tree-based Gradient Boosting และการเติมเต็มตารางกริดแบบเต็มรูปแบบ (Full Cartesian Grid) เพื่อป้องกันความลำเอียงของโมเดล

| `demand_count` | int32 | **[ตัวแปรเป้าหมาย - Target]** จำนวนการเรียกรถที่เกิดขึ้นจริงในชั่วโมงและโซนนั้น |

| `hour_sin`, `hour_cos` | float32 | การแปลงวัฏจักรชั่วโมงในรอบวันให้อยู่ในรูปฟังก์ชันตรีโกณมิติ |
| `dow_sin`, `dow_cos` | float32 | การแปลงวัฏจักรวันในสัปดาห์ |

| `lag_1h`, `lag_2h`, `lag_3h` | float32 | ความต้องการย้อนหลัง 1, 2, 3 ชั่วโมงในโซนเดียวกัน |
| `lag_24h`, `lag_168h` | float32 | ความต้องการเวลาเดียวกันของเมื่อวาน และเวลาเดียวกันของสัปดาห์ก่อน |
| `rolling_mean_3h/6h/24h` | float32 | ค่าเฉลี่ยเคลื่อนที่ย้อนหลัง (คำนวณหลัง `.shift(1)`) |
| `rolling_std_24h` | float32 | ส่วนเบี่ยงเบนมาตรฐานเคลื่อนที่ย้อนหลัง 24 ชั่วโมง |
